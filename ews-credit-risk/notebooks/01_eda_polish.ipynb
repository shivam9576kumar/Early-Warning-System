# 01 — Exploratory Data Analysis: Polish Companies Bankruptcy Dataset
This notebook loads the 5 annual ARFF files from the UCI Polish Companies Bankruptcy Dataset, inspects missing value rates, and documents class imbalance across horizons.


```python
import sys
from pathlib import Path
REPO_ROOT = Path('.').resolve().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from src.data.load_polish import load_all_polish_horizons
df, stats = load_all_polish_horizons()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

stats_df = pd.DataFrame.from_dict(stats, orient='index')
stats_df['horizon_name'] = [f'Horizon {h} ({h} yr prior)' for h in stats_df.index]
stats_df
```

```python
plt.figure(figsize=(8, 4), dpi=150)
sns.barplot(data=stats_df, x='horizon_name', y='bankrupt_pct', palette='Blues_r')
plt.ylabel('Bankrupt % (Class Imbalance)')
plt.title('Bankruptcy Rate by Observation Horizon (Years Before Distress)')
plt.tight_layout()
plt.show()
```
